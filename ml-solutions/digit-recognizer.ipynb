## 1. Import Libraries & Load Data

```python
import kagglehub
import pandas as pd
import xgboost as xgb
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
```

## 2. Preparing dataframes

```python
path = kagglehub.competition_download('digit-recognizer')
train_df = pd.read_csv(f"{path}/train.csv")
train_df.head()
```

```python
test_df = pd.read_csv(f"{path}/test.csv")
test_df.head()
```

## 3. Training a model

```python
X_train = train_df.drop('label', axis=1)
y_train = train_df['label']

X_train_part, X_val, y_train_part, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

model = RandomForestClassifier(random_state=42, n_estimators=200, max_depth=25)

model.fit(X_train_part, y_train_part)
final_preds_val = model.predict(X_val)
final_preds_train = model.predict(X_train_part)
```

## 4. Use of metrics to assess quality

```python
precision = precision_score(y_val, final_preds_val, average='macro') 
recall = recall_score(y_val, final_preds_val, average='macro')
f1 = f1_score(y_val, final_preds_val, average='macro')

print(f"Validation accuracy: {accuracy_score(y_val, final_preds_val):.3f}")
print(f"Training accuracy: {accuracy_score(y_train_part, final_preds_train):.3f}")
print(f"Precision (macro): {precision:.3f}")
print(f"Recall (macro): {recall:.3f}")
print(f"F1-score (macro): {f1:.3f}")
```

## 5. Final predictions and csv

```python
preds = model.predict(test_df)
submission = pd.DataFrame({'ImageId': range(1, len(preds)+1), 'Label': preds})
submission.to_csv('../submissions/digit-recognizer-submission.csv', index=False)
print(submission.head())
```
