## 1. Import Libraries & Load Data

```python
import kagglehub
import pandas as pd
import xgboost as xgb
```

```python
path = kagglehub.competition_download('digit-recognizer')
train_df = pd.read_csv(f"{path}/train.csv")
train_df.head()

X = train_df.drop('label', axis=1)
y = train_df['label']
```

```python
test_df = pd.read_csv(f"{path}/test.csv")
test_df.head()
```

```python
model = xgb.XGBClassifier(
    n_estimators=50,     
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    tree_method='hist',
    n_jobs=-1,             
    random_state=42
)

model.fit(X, y)

preds = model.predict(test_df)
submission = pd.DataFrame({'ImageId': range(1, len(preds)+1), 'Label': preds})
submission.to_csv('../submissions/digit-recognizer-submission.csv', index=False)
```
